```python
str1 = "Hello "
str2 = input("Enter your Name: ")
```

```python
greeting = str1 + " " + str2
print(greeting)
```

```python
str3 = ", welcome to Python programming"
full_greeting = greeting + str3
print(full_greeting)
```

 a. Print the first character of the string

```python
print("a. First character:", full_greeting[0])
```

b. Print the last character of the string

```python
print("b. Last character:", full_greeting[-1])
```

 c. Print the first 5 characters of the string

```python
print("c. First 5 characters:", full_greeting[:5])
```

d. Print the last 11 characters of the string

```python
print("d. Last 11 characters:", full_greeting[-11:])
```

 e. Print the string in reverse

```python
print("e. Reversed string:", full_greeting[::-1])
```

f. Use slicing to print the word "Python" from the existing string

```python
start_idx = full_greeting.find("Python")
print("f. Sliced word:", full_greeting[start_idx : start_idx + 6])
```

3. String Methods

```python
strM = "Python beginner tutorial"
```

a. Convert the sentence to uppercase.

```python
print("a. Uppercase:", strM.upper())
```

 b. Convert the sentence to lowercase

```python
print("b. Lowercase:", strM.lower())
```

c. Use capitalize() to return the sentence to original input form

```python
print("c. Capitalized:", strM.lower().capitalize())
```

d. Count total occurrences of character 't'

```python
print("d. Count of 't':", strM.count('t'))
```

e. Replace all occurrences of "Python" with "Machine Learning"

```python
print("e. Replaced:", strM.replace("Python", "Machine Learning"))
```
